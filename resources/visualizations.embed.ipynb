# Visualisierungen

Diagramme und Plots zu zentralen Workshop-Konzepten

## 1. Kognitive Belastung: Verteilung im Arbeitsgedächtnis

Die Arbeitsgedächtniskapazität ist begrenzt. Entscheidend ist, wie die
drei Arten kognitiver Belastung diesen begrenzten Raum aufteilen.

## 2. Expertise Reversal Effekt

Optimale Unterstützung hängt vom Vorwissen ab. Was Anfängern hilft,
schadet Experten, und umgekehrt. Permanentes KI-Scaffolding verhindert
den Übergang.

## 3. Testing Effekt

Aktives Abrufen führt zu besserem Langzeitlernen als wiederholtes
Studieren, selbst wenn die Gesamtlernzeit gleich bleibt.

## 4. Das Lernparadox

Subjektives Gefühl und tatsächliches Lernen können auseinanderklaffen.
Die metakognitive Falle: Was sich leicht anfühlt, fühlt sich nach Lernen
an, ist es aber oft nicht.

## 5. Offloading vs. Outsourcing: Ein Spektrum

Nicht jede KI-Nutzung wirkt gleich auf das Lernen. Entscheidend ist, ob
die lernrelevante kognitive Verarbeitung erhalten bleibt oder eliminiert
wird.

## 6. Arbeitsgedächtnis-Flaschenhals

Informationen müssen durch das begrenzte Arbeitsgedächtnis, bevor sie im
Langzeitgedächtnis verankert werden können.

## 7. Das Evaluationsparadox

Das Evaluationsparadox als zirkuläre Abhängigkeit: Um KI-Outputs zu
bewerten, braucht man Fachkompetenz, die erst durch eigenes Lernen
entsteht.

**Kontrast: Experten vs. Anfänger**

## 8. Versuch-dann-Prüfe Zyklus

Das zentrale Designprinzip für KI im Lernkontext: Studierende generieren
zuerst eigene Lösungen, dann vergleichen sie mit KI-Output.

## 9. Das 5-Fragen-Framework

Der vollständige Entscheidungsbaum für die Aufgabenanalyse im
KI-Zeitalter.

## 10. Das 5-Fragen-Framework (Zusammenfassung)

Die erweiterte Version des Entscheidungsbaums mit explizitem “Keine KI
nötig”-Pfad.
